Wordnet Lemmatizer

Lemmatization technique is like stemming. The output we will get after lemmatization is called lemma', which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet Corpus Reader class to find a lemma. Let us understand it with an example 

```python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
import nltk
nltk.download('wordnet')
```

```python
'''POS- Noun-n
verb-v
adjective-a
adverb-r'''
lemmatizer.lemmatize("going", pos='a')
```

```python
lemmatizer.lemmatize("goes", pos='v')
```

```python
lemmatizer.lemmatize("fairly", pos='v'), lemmatizer.lemmatize("sportingly")
```

```python
words=["eating",'eats','eaten','writes','programs','programming',"writing","history",'finally','finalize']
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos="v"))
```
