STEMMING
Stemming is the process of reducing inflected (or sometimes derived) words to their word stem, base or root form—generally a written word form.stemming is important in natural language understanding (NLU) and natural language processing(NLP).

RegexpStemmer class

NLTK has Regexp Stemmer class with the help of which we can easily implement Regular

Expression Stemmer algorithms. It basically takes a single regular expression and removes any prefix or suffix that matches the expression. Let us see an example

RegexpStemmer class

NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basically takes a single regular expression and removes any prefix or suffix that matches the expression.

```python
## Classification Problem
## Comments of product is a positive review or negative review
## Reviews----> eating, eat, eaten [going, gone, goes]--->go

```

```python
#PorterStemmer
from nltk.stem import PorterStemmer
stemming=PorterStemmer()
words=["eating",'eats','eaten','writes','programs','programming',"writing","history",'finally','finalize']
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem("congratulations")
```

```python
stemming.stem("sitting")
```

```python
from nltk.stem import RegexpStemmer
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

```python
#Snowball Stemmer
from nltk.stem import SnowballStemmer
snowballsstemmer=SnowballStemmer('english')
```

```python
for word in words:
    print (word+"--->" +snowballsstemmer.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
snowballsstemmer.stem("fairly"),snowballsstemmer.stem("sportingly")
```

```python
snowballsstemmer.stem("goes")
```
