1.PREPROCESSING AND CLEANING

```python
import pandas as pd

# Switch sep to '\t' so pandas ignores the commas inside the text messages
messages = pd.read_csv("Spamclassifier-master.csv", sep=',', names=['label', 'message'])
```

```python
messages
```

2.BOW AND TF-IDF(SENTENCES->VECTORS){PREVENTING DATA LEAKAGE}

```python
import re
import nltk
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
ps=PorterStemmer()
```

```python
corpus = []
stop_words = set(stopwords.words('english'))

for i in range(0, len(messages)):
   
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if word not in stop_words]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus
```

3.TRAIN TEST SPLITS

```python
#create the bag of words model with ngram
from sklearn.feature_extraction.text import CountVectorizer
# for binary Bow enable binary=True
cv=CountVectorizer(max_features=2500,binary=True,ngram_range=(1,2))
x=cv.fit_transform(corpus).toarray()
```

```python
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000,formatter=dict(float=lambda x:"%.3g" % x))
x
```

```python
##output features
y=pd.get_dummies(messages['label'])
```

```python
y=y.iloc[:,0].values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.20)
```

4.TRAINED OUR MODELS

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
spam_detect_model=MultinomialNB().fit(X_train,y_train)
```

```python
y_pred=spam_detect_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

USING TF-IDF FOR TRAIN TEST

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(max_features=100)
x=tfidf.fit_transform(corpus).toarray()
```

```python
x
```

```python
tfidf=TfidfVectorizer(max_features=100,ngram_range=(2,2))
x=tfidf.fit_transform(corpus).toarray()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.20)
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_detect_model=MultinomialNB().fit(X_train,y_train)
```

```python
y_pred=spam_detect_model.predict(X_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```
