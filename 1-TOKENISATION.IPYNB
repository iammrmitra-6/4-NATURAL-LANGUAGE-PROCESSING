```python
import nltk
nltk.download('punkt')
```

```python
from nltk.tokenize import sent_tokenize
```

```python
# Define your text corpus
corpus = "Hello there! Welcome to natural language processing. Isn't NLTK great for this?"
```

```python
# Explicitly import the sentence tokenizer
from nltk.tokenize import sent_tokenize
# Run the tokenization
sentences = sent_tokenize(corpus)
print(sentences)
```

```python
sent_tokenize(corpus)
```

```python
type(sentences)
```

```python
for s in sentences:
    print(s)
```
