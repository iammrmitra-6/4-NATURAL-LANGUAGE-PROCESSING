```python
import nltk
import ssl

# 1. Bypass potential local SSL network limitations
try:
    _create_unverified_https_context = ssl._create_unverified_context
except AttributeError:
    pass
else:
    ssl._create_default_https_context = _create_unverified_https_context

# 2. Download the modern _tab version of the chunker
print("Downloading required NLTK data structure...")
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
print("Downloads ready!")

# 3. Process data
tag_elements = [('Taj', 'NNP'), ('Mahal', 'NNP'), ('is', 'VBZ'), ('beautiful', 'JJ')]

# 4. Parse and render GUI tree
chunked_tree = nltk.ne_chunk(tag_elements)
chunked_tree.draw()
```
